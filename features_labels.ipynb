```python
# diagnosis is the label (y)
import pandas as pd
import numpy as np
import kagglehub
from sklearn.model_selection import train_test_split

from sklearn.svm import SVC

# StandardScaler (and pipeline) is used to avoid the ConvergenceWarning: failed to converge
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# Download to /kaggle/input/breast-cancer-wisconsin-data
path = kagglehub.dataset_download("uciml/breast-cancer-wisconsin-data")

df = pd.read_csv("/kaggle/input/breast-cancer-wisconsin-data/data.csv")
# input sanity check: NaN drop
df = df.dropna(axis=1, how="all")

# 'id','diagnosis' cols to drop
X = np.array(df.drop(columns=['id','diagnosis'], axis = 1))
# labels
y = np.array(df['diagnosis'])

# just to check the a row
with np.printoptions(suppress=True, precision=3):
    print(X[1],y[1])
```

```python
[X_train, X_test, y_train, y_test] = train_test_split(X, y, test_size = 0.20, random_state = 45564)
print(type(X_train), X_train.shape)  # N-dim array
```

```python
Classifier = SVC(kernel='linear')
model = Classifier.fit(X_train, y_train)
accu = model.score(X_test, y_test)
accu
```

```python
Classifier = LogisticRegression()
# model = Classifier.fit(X_train, y_train)
model = make_pipeline(StandardScaler(), LogisticRegression())
model.fit(X_train, y_train)
# alternatively, scale manually
# scaler = StandardScaler().fit(X_train)
# X_train_s = scaler.transform(X_train)
# X_test_s = scaler.transform(X_test)

accu = model.score(X_test, y_test)
accu
```

```python
import pickle

# save the model (and share with the community)
pickle.dump(model, open("logregr_scaled.m","wb"))

# or load it...
model_loaded = pickle.load(open("logregr_scaled.m","rb"))
accu = model.score(X_test, y_test)
accu_loaded = model_loaded.score(X_test, y_test)
print ('original:', model.named_steps,'\nloaded:' ,model_loaded.named_steps,'\noriginal accu:',
       accu,'\nloaded accu:' ,accu_loaded)
```
